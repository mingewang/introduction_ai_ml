### Problem 1: Plot Training Loss

### Problem 2: Detect Overfitting

**Given:**
```python
train_loss = [2.5, 2.0, 1.6, 1.3, 1.1, 0.9]
val_loss   = [2.6, 2.2, 2.0, 2.1, 2.3, 2.6]
```

**Task:** Plot both curves on the same axes.

- Add a legend identifying the two lines.  
- Label axes and title **"Training vs Validation Loss"**.  
- Assign figure and axes to `fig2` and `ax2`.  
- Set `overfitting_epoch` to the epoch index **after which** overfitting begins (integer).


````python
# TODO: Write your code for Problem 2 here
train_loss = [2.5, 2.0, 1.6, 1.3, 1.1, 0.9]
val_loss   = [2.6, 2.2, 2.0, 2.1, 2.3, 2.6]
epochs = range(len(train_loss))

fig2, ax2 = plt.subplots()
ax2.plot(epochs, train_loss, 'b-o', label='Train Loss')
ax2.plot(epochs, val_loss, 'r-s', label='Validation Loss')
ax2.set_xlabel("Epoch")
ax2.set_ylabel("Loss")
ax2.set_title("Training vs Validation Loss")
ax2.legend()
plt.show()

# TODO: Set this to the epoch index after which overfitting begins
overfitting_epoch = 2
````

### Problem 3: Two-Axes Figure

**Given:**
```python
epochs      = [0, 1, 2, 3, 4, 5]
loss        = [3.0, 2.4, 1.9, 1.6, 1.4, 1.3]
accuracy    = [0.60, 0.72, 0.81, 0.85, 0.88, 0.90]
```

**Task:** Create a figure with **two y-axes** (left and right).  
- Left y-axis → loss (blue, solid line)  
- Right y-axis → accuracy (red, dashed line)  
- x-axis label: **"Epoch"**, left y-label: **"Loss"**, right y-label: **"Accuracy"**  
- Title: **"Loss & Accuracy over Epochs"**  
- Assign the figure and **left** axes to `fig3` and `ax3`.


````python
# TODO: Write your code for Problem 3 here
epochs   = [0, 1, 2, 3, 4, 5]
loss     = [3.0, 2.4, 1.9, 1.6, 1.4, 1.3]
accuracy = [0.60, 0.72, 0.81, 0.85, 0.88, 0.90]

fig3, ax3 = plt.subplots()
ax3.plot(epochs, loss, 'b-', label='Loss')
ax3.set_xlabel("Epoch")
ax3.set_ylabel("Loss", color='b')

ax3b = ax3.twinx()
ax3b.plot(epochs, accuracy, 'r--', label='Accuracy')
ax3b.set_ylabel("Accuracy", color='r')

ax3.set_title("Loss & Accuracy over Epochs")
plt.show()
````

### Task 1: Simple Bar Chart

````python
# TODO: Write your code for Bar Task 1 here
fruits = ['Apple', 'Banana', 'Orange', 'Grapes']
counts = [120, 80, 150, 50]

fig_b1, ax_b1 = plt.subplots()
bars = ax_b1.bar(fruits, counts)
ax_b1.set_xlabel("Fruit")
ax_b1.set_ylabel("Count")
ax_b1.set_title("Fruit Counts")
for bar, count in zip(bars, counts):
    ax_b1.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 2,
               str(count), ha='center', va='bottom')
plt.show()
````

### Task 1: Small Dataset Correlation

**Task:**  
1. Create a dataset with **4 features** and **10 rows** of random integers (use `np.random.randint`, seed=42).  
2. Compute the correlation matrix with `np.corrcoef`.  
3. Plot a heatmap using `plt.matshow` and add a colorbar.  
4. Assign the figure to `fig_c1`.


````python
# TODO: Write your code for Correlation Task 1 here
np.random.seed(42)
data = np.random.randint(0, 100, size=(10, 4))
corr = np.corrcoef(data.T)

fig_c1 = plt.figure()
plt.matshow(corr, fignum=0)
plt.colorbar()
plt.title("Correlation Matrix")
plt.show()
````

````python
# ---- Validation: Correlation Task 1 ----
assert 'fig_c1' in dir(), "❌ Missing fig_c1 variable"
assert 'corr' in dir(), "❌ Missing corr variable"
assert corr.shape == (4, 4), f"❌ Correlation matrix should be (4,4), got {corr.shape}"
assert np.allclose(corr.diagonal(), 1.0), "❌ Diagonal should be all 1s"
assert np.allclose(corr, corr.T), "❌ Correlation matrix should be symmetric"
print("✅ Correlation Task 1 passed!")
````

### Task 4: Real Dataset

**Task:**  
1. Create a synthetic CSV-like dataset with 5–6 numerical features (use `np.random.randn`).  
2. Compute the correlation matrix.  
3. Plot a heatmap (seaborn).  
4. Identify which two features are **most strongly correlated** (highest absolute correlation aside from diagonal).  
   Set `strongest_pair` to a tuple like `(0, 2)` (feature indices).

> If you prefer, you can instead load a real CSV from disk. Just ensure it has numerical columns.


````python
# TODO: Write your code for Correlation Task 4 here
np.random.seed(123)
dataset = np.random.randn(50, 6)
# Introduce correlation between col 1 and col 4
dataset[:, 4] = dataset[:, 1] * 0.8 + np.random.randn(50) * 0.6

corr_real = np.corrcoef(dataset.T)

fig_c4, ax_c4 = plt.subplots(figsize=(7, 6))
sns.heatmap(corr_real, annot=True, cmap='coolwarm', ax=ax_c4,
            xticklabels=[f'F{i}' for i in range(6)],
            yticklabels=[f'F{i}' for i in range(6)])
ax_c4.set_title("Real Dataset Correlation")
plt.show()

# TODO: Which two feature indices have the strongest correlation?
# (excluding diagonal)
strongest_pair = (1, 4)
````

````python
# ---- Validation: Correlation Task 4 ----
assert 'fig_c4' in dir(), "❌ Missing fig_c4 variable"
assert 'corr_real' in dir(), "❌ Missing corr_real variable"
assert corr_real.shape[0] >= 5, f"❌ Expected at least 5 features, got {corr_real.shape[0]}"
# Check strongest_pair
i, j = strongest_pair
off_diag = np.abs(corr_real - np.eye(corr_real.shape[0]))
assert off_diag[i, j] == off_diag.max(), \
    f"❌ Pair {strongest_pair} (corr={corr_real[i,j]:.3f}) is not the strongest correlated pair"
print(f"✅ Correlation Task 4 passed! Strongest pair: features {strongest_pair} with correlation {corr_real[i,j]:.3f}")
````
