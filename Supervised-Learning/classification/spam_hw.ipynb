## Bayes' Theorem

$$ P(C|X) = \frac{P(X|C) \cdot P(C)}{P(X)} $$

- $P(C|X)$: posterior probability
- $P(X|C)$: likelihood
- $P(C)$: prior probability
- $P(X)$: evidence (constant across classes, can be ignored for comparison)

## Naive Bayes Assumption

Features are **conditionally independent** given the class:

$$ P(X|C) = P(X_1|C) \cdot P(X_2|C) \cdot ... \cdot P(X_n|C) $$

So for classification we compare:

$$ P(C|X) \propto P(C) \cdot \prod_{i=1}^{n} P(X_i|C) $$

## Advantages of Naive Bayes

1. **Simple and Fast**: computationally efficient, requires simple probability calculations.
2. **Works Well with High-Dimensional Data**: effective in text classification with many features.
3. **Handles Missing Data**: missing features can be ignored during classification.

## Disadvantages

1. **Feature Independence Assumption**: performance degrades if features are strongly correlated.
2. **Poor Performance with Small Data**: probability estimates can be inaccurate with too little data.
