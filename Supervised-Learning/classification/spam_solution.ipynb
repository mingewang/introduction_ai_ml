# Naive Bayes Classifier: Detailed Explanation with Example

Naive Bayes is a probabilistic classification algorithm based on **Bayes' Theorem**, which assumes that the features (predictors) are independent given the class label. Despite the strong assumption of feature independence (which is often violated in real-world datasets), Naive Bayes can work well in practice, especially in text classification tasks, like spam detection, sentiment analysis, etc.

## Bayes' Theorem

Bayes' Theorem gives the probability of a class $C$ given the feature values $X = (X_1, X_2, ..., X_n)$ using the following formula:

$$ P(C|X) = \frac{P(X|C) \cdot P(C)}{P(X)} $$

Where:
- $P(C|X)$ is the **posterior probability**, the probability of class $C$ given the features $X$.
- $P(X|C)$ is the **likelihood**, the probability of observing features $X$ given the class $C$.
- $P(C)$ is the **prior probability** of class $C$, which is the probability of class $C$ occurring before observing any features.
- $P(X)$ is the **evidence**, the total probability of observing the features $X$ (often not computed directly as it is constant for all classes).

## Naive Bayes Assumption

Naive Bayes makes the naive assumption that all features are **conditionally independent** given the class label. That is, for each class $C$ and feature $X_1, X_2, ..., X_n$:

$$ P(X|C) = P(X_1|C) \cdot P(X_2|C) \cdot ... \cdot P(X_n|C) $$

Thus, we can rewrite Bayes' Theorem for Naive Bayes as:

$$ P(C|X) = \frac{P(C) \cdot P(X_1|C) \cdot P(X_2|C) \cdot ... \cdot P(X_n|C)}{P(X)} $$

Since $P(X)$ is constant for all classes, we can focus on the numerator and ignore the denominator when comparing probabilities across classes:

$$ P(C|X) \propto P(C) \cdot \prod_{i=1}^{n} P(X_i|C) $$

## Steps in Naive Bayes Classification

1. **Estimate Prior Probabilities**: Calculate the prior probability $P(C)$ for each class based on the training data. This is simply the relative frequency of each class in the dataset.

2. **Estimate Likelihood**: For each feature $X_i$, calculate $P(X_i|C)$, which is the likelihood of the feature $X_i$ given class $C$. This is done for each feature in the dataset.

3. **Apply Bayes' Theorem**: For a new sample, calculate the posterior probability for each class $C$ using the formula above.

4. **Classification**: The class with the highest posterior probability is chosen as the predicted class.

## Example: Email Classification (Spam vs. Ham)

Let's illustrate how Naive Bayes works with a simple example of classifying emails as **Spam** or **Ham** (non-spam).

### Step 1: Training Data

Suppose we have the following training dataset:

| Email ID | Word 1 (Free) | Word 2 (Money) | Class |
|----------|---------------|----------------|-------|
| 1        | Yes           | No             | Spam  |
| 2        | No            | Yes            | Spam  |
| 3        | Yes           | Yes            | Ham   |
| 4        | No            | No             | Ham   |
| 5        | Yes           | No             | Ham   |

From this data, we want to calculate the prior probabilities and the likelihoods.

```python
import pandas as pd

# Create the training dataset
data = {
    'Free': ['Yes', 'No', 'Yes', 'No', 'Yes'],
    'Money': ['No', 'Yes', 'Yes', 'No', 'No'],
    'Class': ['Spam', 'Spam', 'Ham', 'Ham', 'Ham']
}
df = pd.DataFrame(data, index=[1, 2, 3, 4, 5])
df.index.name = 'Email ID'
df
```

### Step 2: Calculate Prior Probabilities

The prior probability of each class is the relative frequency of each class in the training set.

$$ P(Spam) = \frac{2}{5} = 0.4 $$

$$ P(Ham) = \frac{3}{5} = 0.6 $$

```python
# Calculate prior probabilities
prior_spam = len(df[df['Class'] == 'Spam']) / len(df)
prior_ham = len(df[df['Class'] == 'Ham']) / len(df)

print(f"P(Spam) = {prior_spam}")
print(f"P(Ham) = {prior_ham}")
```

### Step 3: Calculate Likelihoods

We now calculate the likelihood for each feature (word) given the class.

**For Spam:**

$ P(Free = Yes \mid Spam) = \frac{2}{2} = 1 $  (Both spam emails contain the word "Free")

$ P(Money = Yes \mid Spam) = \frac{1}{2} = 0.5 $ (One spam email contains the word "Money", one does not)

**For Ham:**

$ P(Free = Yes \mid Ham) = \frac{2}{3} $ (Two ham emails contain the word "Free")

$ P(Money = Yes \mid Ham) = \frac{1}{3} $ (One ham email contains the word "Money", two do not)

```python
# Calculate likelihoods
spam_df = df[df['Class'] == 'Spam']
ham_df = df[df['Class'] == 'Ham']

# For Spam
p_free_yes_spam = len(spam_df[spam_df['Free'] == 'Yes']) / len(spam_df)
p_money_yes_spam = len(spam_df[spam_df['Money'] == 'Yes']) / len(spam_df)

# For Ham
p_free_yes_ham = len(ham_df[ham_df['Free'] == 'Yes']) / len(ham_df)
p_money_yes_ham = len(ham_df[ham_df['Money'] == 'Yes']) / len(ham_df)

print("=== Likelihoods for Spam ===")
print(f"P(Free=Yes | Spam) = {p_free_yes_spam}")
print(f"P(Money=Yes | Spam) = {p_money_yes_spam}")

print("\n=== Likelihoods for Ham ===")
print(f"P(Free=Yes | Ham) = {p_free_yes_ham}")
print(f"P(Money=Yes | Ham) = {p_money_yes_ham}")
```

### Step 4: Calculate Posterior Probabilities

Now, suppose we receive a new email with the words "Free" and "Money". We want to classify this email.

**For Spam:**

$$ P(Spam \mid Free=Yes, Money=Yes) \propto P(Spam) \cdot P(Free=Yes \mid Spam) \cdot P(Money=Yes \mid Spam) $$

$$ P(Spam \mid Free=Yes, Money=Yes) \propto 0.4 \cdot 1 \cdot 0.5 = 0.2 $$

**For Ham:**

$$ P(Ham \mid Free=Yes, Money=Yes) \propto P(Ham) \cdot P(Free=Yes \mid Ham) \cdot P(Money=Yes \mid Ham) $$

$$ P(Ham \mid Free=Yes, Money=Yes) \propto 0.6 \cdot \frac{2}{3} \cdot \frac{1}{3} = 0.1333 $$

```python
# New email: Free=Yes, Money=Yes
posterior_spam = prior_spam * p_free_yes_spam * p_money_yes_spam
posterior_ham = prior_ham * p_free_yes_ham * p_money_yes_ham

print(f"Posterior probability for Spam: {posterior_spam:.4f}")
print(f"Posterior probability for Ham: {posterior_ham:.4f}")
```

### Step 5: Classification

We compare the posterior probabilities for each class:

- $P(Spam \mid Free=Yes, Money=Yes) = 0.2$
- $P(Ham \mid Free=Yes, Money=Yes) = 0.1333$

Since $P(Spam \mid Free=Yes, Money=Yes) > P(Ham \mid Free=Yes, Money=Yes)$, we classify the email as **Spam**.

```python
# Classification decision
if posterior_spam > posterior_ham:
    print("Prediction: SPAM")
else:
    print("Prediction: HAM")
```

## Advantages of Naive Bayes

1. **Simple and Fast**: Naive Bayes is computationally efficient because it requires only a simple calculation of probabilities for each feature in the dataset.

2. **Works Well with High-Dimensional Data**: It is particularly effective in text classification problems, where the number of features (words) can be large.

3. **Handles Missing Data**: Naive Bayes can handle missing features by simply ignoring them during the classification process.

## Disadvantages of Naive Bayes

1. **Feature Independence Assumption**: The biggest limitation is the independence assumption. If the features are strongly correlated, the performance can degrade.

2. **Poor Performance with Small Data**: If the training data is too small, the estimates of the probabilities can be inaccurate.

## Conclusion

Naive Bayes is a powerful and simple classifier that can perform surprisingly well, especially in high-dimensional spaces like text classification, despite its strong independence assumption. It is widely used for tasks like spam detection, sentiment analysis, and document classification.
