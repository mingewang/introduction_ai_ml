```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## 2. Loading Example Datasets

Seaborn includes several built-in datasets. Let's load the `tips` dataset.

```python
# Load the 'tips' dataset
tips = sns.load_dataset('tips')
print(tips.head())
print()
print(tips.describe())
```

## 3. Basic Plot Types

### 3.1 Scatter Plot
Visualize relationships between two numeric variables.

```python
# Scatter plot: Total bill vs Tip
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='day', style='time', size='size', palette='deep')
plt.title('Scatter Plot: Total Bill vs Tip')
plt.show()
```

### 3.2 Line Plot
Display trends over time or across ordered data.

```python
# Line plot of sine wave
x = np.linspace(0, 10, 100)
y = np.sin(x)

sns.lineplot(x=x, y=y, color='blue', label='Sine Wave')
plt.title('Line Plot: Sine Wave')
plt.legend()
plt.show()
```

### 3.3 Histogram
Understand the distribution of a single variable.

```python
# Histogram: Distribution of Total Bill
sns.histplot(data=tips, x='total_bill', kde=True, bins=20, color='orange')
plt.title('Histogram: Distribution of Total Bill')
plt.show()
```

### 3.4 Box Plot
Visualize the distribution of data based on statistical summaries.

```python
# Box plot: Total Bill by Day
sns.boxplot(data=tips, x='day', y='total_bill', palette='Set2')
plt.title('Box Plot: Total Bill by Day')
plt.show()
```

### 3.5 Violin Plot
Combine a box plot with a kernel density estimate.

```python
# Violin plot: Tip by Day, split by sex
sns.violinplot(data=tips, x='day', y='tip', hue='sex', split=True, palette='muted')
plt.title('Violin Plot: Tip by Day')
plt.show()
```

## 4. Statistical Visualizations

### 4.1 Pair Plot
Visualize pairwise relationships in a dataset.

```python
# Pair plot of numerical columns
sns.pairplot(data=tips, hue='sex', palette='coolwarm')
plt.show()
```

### 4.2 Heatmap
Display correlations or other matrix data.

```python
# Correlation heatmap
corr = tips[['total_bill', 'tip', 'size']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```

## 5. Customizing Seaborn Visualizations

### 5.1 Changing Themes
Seaborn provides built-in themes to style plots.

```python
# Set dark theme
sns.set_theme(style="darkgrid")
sns.histplot(data=tips, x='total_bill', kde=True, color='green')
plt.title('Styled Histogram (Dark Grid)')
plt.show()
```

### 5.2 Adjusting Figure Size

```python
# Custom figure size
plt.figure(figsize=(10, 6))
sns.boxplot(data=tips, x='day', y='total_bill', palette='Set3')
plt.title('Box Plot with Custom Figure Size')
plt.show()
```

## 6. Combining Seaborn with Matplotlib

Seaborn integrates seamlessly with Matplotlib for advanced customization.

```python
# Overlayed histograms
sns.histplot(data=tips, x='total_bill', kde=True, color='blue', alpha=0.5, label='Total Bill')
sns.histplot(data=tips, x='tip', kde=True, color='red', alpha=0.5, label='Tip')
plt.legend()
plt.title('Overlayed Histograms')
plt.show()
```

## 7. Advanced Seaborn Plots

### 7.1 Joint Plot
Displays a scatter plot with histograms on the axes.

```python
# Joint plot with regression line
sns.jointplot(data=tips, x='total_bill', y='tip', kind='reg')
plt.show()
```

### 7.2 Facet Grid
Create multiple plots based on subsets of data.

```python
# Facet grid: scatter plots by day
g = sns.FacetGrid(data=tips, col='day', hue='sex', height=4, aspect=0.8)
g.map(sns.scatterplot, 'total_bill', 'tip').add_legend()
plt.show()
```
